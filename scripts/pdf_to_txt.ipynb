````python
from pathlib import Path

import pdfplumber          # đọc searchable PDF (PDF-text)
import fitz                # từ pymupdf, đọc PDF/ảnh
from PIL import Image      # Pillow
import pytesseract         # OCR
````

````python
TESSERACT_LANG = "vie"   # hoặc "vie+eng" nếu muốn nhận cả tiếng Anh
````

````python
def is_searchable_pdf(pdf_path: Path, min_chars: int = 50, max_pages_check: int = 5) -> bool:
    """
    Kiểm tra PDF có phải searchable (PDF-text) không.
    Ý tưởng: đọc vài trang đầu bằng pdfplumber, nếu lấy được đủ ký tự text thì coi là searchable.
    """
    with pdfplumber.open(str(pdf_path)) as doc:
        total_chars = 0
        num_pages = min(len(doc.pages), max_pages_check)

        for i in range(num_pages):
            page = doc.pages[i]
            text = page.extract_text() or ""
            total_chars += len(text.strip())
            if total_chars >= min_chars:
                return True

    return False
````

````python
def extract_text_searchable(pdf_path: Path) -> str:
    """
    Trích văn bản từ searchable PDF bằng pdfplumber.
    """
    print(f"[INFO] Extract text (searchable) from: {pdf_path.name}")
    all_text = []

    with pdfplumber.open(str(pdf_path)) as doc:
        for i, page in enumerate(doc.pages):
            text = page.extract_text() or ""
            all_text.append(text)
            if (i + 1) % 10 == 0:
                print(f"  - Đã xử lý {i+1}/{len(doc.pages)} trang...")

    full_text = "\n\n".join(all_text)
    print(f"[OK] Hoàn thành extract searchable PDF, tổng {len(full_text)} ký tự.")
    return full_text
````

````python
def ocr_scanned_pdf(pdf_path: Path, dpi: int = 300) -> str:
    """
    Chuyển từng trang PDF thành ảnh, sau đó OCR bằng Tesseract để lấy text.
    """
    print(f"[INFO] OCR scanned PDF: {pdf_path.name}")
    doc = fitz.open(str(pdf_path))
    all_text = []

    for page_num in range(len(doc)):
        page = doc.load_page(page_num)
        pix = page.get_pixmap(dpi=dpi)

        # Chuyển Pixmap của PyMuPDF sang PIL Image
        mode = "RGB" if pix.n < 4 else "RGBA"
        img = Image.frombytes(mode, [pix.width, pix.height], pix.samples)

        # OCR bằng Tesseract
        text = pytesseract.image_to_string(img, lang=TESSERACT_LANG)
        all_text.append(text)

        if (page_num + 1) % 5 == 0 or page_num == len(doc) - 1:
            print(f"  - Đã OCR {page_num+1}/{len(doc)} trang...")

    full_text = "\n\n".join(all_text)
    print(f"[OK] Hoàn thành OCR scanned PDF, tổng {len(full_text)} ký tự.")
    return full_text
````

````python
def pdf_to_txt_pipeline(pdf_path: Path, txt_path: Path):
    """
    Quy trình Bước 1 cho VanHoaVaDuLichVN.pdf:
    - Kiểm tra PDF searchable hay scanned.
    - Nếu searchable: dùng pdfplumber extract text.
    - Nếu scanned: dùng PyMuPDF + Tesseract OCR để lấy text.
    - Ghi toàn bộ text ra file .txt.
    """
    print(f"========== BẮT ĐẦU XỬ LÝ: {pdf_path.name} ==========")

    if not pdf_path.exists():
        raise FileNotFoundError(f"Không tìm thấy file PDF: {pdf_path}")

    # 1. Kiểm tra searchable?
    print("[STEP] Kiểm tra loại PDF (searchable hay scanned)...")
    searchable = is_searchable_pdf(pdf_path)
    print(f"[INFO] Searchable PDF?  {'YES' if searchable else 'NO'}")

    # 2. Trích xuất text theo nhánh tương ứng
    if searchable:
        full_text = extract_text_searchable(pdf_path)
    else:
        full_text = ocr_scanned_pdf(pdf_path, dpi=300)

    # 3. Ghi ra file txt
    txt_path.parent.mkdir(parents=True, exist_ok=True)
    with open(txt_path, "w", encoding="utf-8") as f:
        f.write(full_text)

    print(f"[DONE] Đã lưu văn bản thuần vào: {txt_path} (số ký tự: {len(full_text)})")

    return full_text
````

````python
from pathlib import Path
from typing import List

txt_path = Path("VanHoaVaDuLichVN.txt")

if not txt_path.exists():
    raise FileNotFoundError(f"Không thấy file {txt_path}, cần chạy lại Bước 1 (PDF -> TXT).")

with open(txt_path, "r", encoding="utf-8") as f:
    full_text = f.read()

print("Số ký tự:", len(full_text))
print("\n--- 500 ký tự đầu ---\n")
print(full_text[:500])

def split_text_into_chunks(text: str, chunk_size: int = 1500, overlap: int = 200) -> List[str]:
    """
    Chia text thành nhiều đoạn (chunk) theo số từ.
    chunk_size: số từ mỗi chunk.
    overlap: số từ trùng giữa 2 chunk để giữ ngữ cảnh.
    """
    words = text.split()
    chunks = []
    i = 0
    while i < len(words):
        chunk_words = words[i : i + chunk_size]
        chunks.append(" ".join(chunk_words))
        if i + chunk_size >= len(words):
            break
        i += chunk_size - overlap
    return chunks

chunks = split_text_into_chunks(full_text, chunk_size=1500, overlap=200)
print(f"\nĐã chia thành {len(chunks)} chunks.")
print("\n--- 300 ký tự đầu chunk 1 ---\n")
print(chunks[0][:300])
````

````python
import json
import re

SYSTEM_PROMPT = """
Bạn là mô hình trích xuất thông tin cho lĩnh vực Văn hoá – Du lịch – Giao thông Việt Nam.

Nhiệm vụ:
- Đọc đoạn văn bản tiếng Việt trong tài liệu "Văn hoá và Du lịch Việt Nam".
- Trích xuất các THỰC THỂ (entities) và SỰ KIỆN (events) quan trọng.

ĐẦU RA:
Trả về DUY NHẤT một đối tượng JSON:

{
  "entities": [
    {
      "id": "ent_1",
      "name": "tên thực thể",
      "type": "place | cultural_site | route | festival | organization | person | other",
      "location": "địa điểm (tỉnh/thành, khu vực) nếu có, hoặc null",
      "time": "thời gian liên quan nếu có, hoặc null",
      "description": "mô tả ngắn gọn",
      "tags": ["từ khoá", "liên quan"]
    }
  ],
  "events": [
    {
      "id": "ev_1",
      "name": "tên sự kiện",
      "time": "thời gian diễn ra nếu có, hoặc null",
      "location": "địa điểm diễn ra nếu có, hoặc null",
      "participants": ["danh sách id ent_* tham gia nếu biết"],
      "description": "mô tả ngắn gọn",
      "tags": ["từ khoá", "liên quan"]
    }
  ]
}

Quy tắc:
- Nếu không tìm thấy gì, trả về {"entities": [], "events": []}.
- KHÔNG thêm trường ngoài các trường trên.
- KHÔNG in giải thích, KHÔNG in lại đề bài, KHÔNG dùng ```json.
- Chỉ in một object JSON hợp lệ.
"""
````

````python
def extract_json_with_gemini(text_chunk: str) -> dict:
    """
    Gửi 1 chunk văn bản lên Gemini, nhận về 1 object JSON theo schema.
    """
    prompt = f"""{SYSTEM_PROMPT.strip()}

Đoạn văn bản:

\"\"\"{text_chunk}\"\"\"

Hãy trích xuất các thực thể và sự kiện theo đúng yêu cầu trên.
Nhớ: CHỈ TRẢ LỜI MỘT OBJECT JSON.
"""

    response = client.models.generate_content(
        model="gemini-2.5-flash",      # hoặc "gemini-1.5-flash" / model khác nếu bạn muốn
        contents=prompt,
        config={
            "response_mime_type": "application/json"  # bắt buộc trả JSON
        },
    )

    raw = response.text.strip()

    # Thử parse trực tiếp
    try:
        data = json.loads(raw)
    except Exception:
        # Nếu model lỡ sinh thêm chữ, cố gắng tách đoạn {...}
        candidates = re.findall(r"\{[\s\S]*\}", raw)
        data = {"entities": [], "events": []}
        for cand in reversed(candidates):  # ưu tiên đoạn cuối
            try:
                obj = json.loads(cand)
                if isinstance(obj, dict) and "entities" in obj and "events" in obj:
                    data = obj
                    break
            except Exception:
                continue

    # Đảm bảo luôn có trường entities & events dạng list
    if "entities" not in data or not isinstance(data["entities"], list):
        data["entities"] = []
    if "events" not in data or not isinstance(data["events"], list):
        data["events"] = []

    return data
````

````python
def validate_record_structure(record: dict) -> bool:
    """
    Kiểm tra tự động (Automation Test):
    - Có 'entities' và 'events'
    - Mỗi entity có id, name, type, description
    - Mỗi event có id, name, description
    """
    if not isinstance(record, dict):
        return False
    if "entities" not in record or "events" not in record:
        return False
    if not isinstance(record["entities"], list) or not isinstance(record["events"], list):
        return False

    for ent in record["entities"]:
        if not all(k in ent for k in ["id", "name", "type", "description"]):
            return False

    for ev in record["events"]:
        if not all(k in ev for k in ["id", "name", "description"]):
            return False

    return True


raw_records = []       # Raw Data JSON mỗi chunk
final_entities = []    # Final Data: entities hợp lệ
final_events = []      # Final Data: events hợp lệ
invalid_records = []   # JSON lỗi để Tester xem

num_chunks = min(10, len(chunks))   # lúc đầu demo 10 chunks, OK rồi thì đổi thành len(chunks)

for i in range(num_chunks):
    print(f"=== Xử lý chunk {i+1}/{num_chunks} ===")
    record = extract_json_with_gemini(chunks[i])
    raw_records.append(record)

    if validate_record_structure(record):
        print("  -> JSON HỢP LỆ (cho vào Final Data).")
        final_entities.extend(record.get("entities", []))
        final_events.extend(record.get("events", []))
    else:
        print("  -> JSON KHÔNG HỢP LỆ (đưa vào Invalid để Tester chỉnh).")
        invalid_records.append(record)

print("\n[SUMMARY]")
print("  Raw records:", len(raw_records))
print("  Entities hợp lệ:", len(final_entities))
print("  Events hợp lệ:", len(final_events))
print("  Records invalid:", len(invalid_records))
````

````python
def create_qa_from_entity(ent: dict, max_questions: int = 5):
    """
    Tạo list Q&A cho 1 thực thể (điểm du lịch / di sản / tuyến, ...),
    tập trung vào góc nhìn khách du lịch.
    ent: {id, name, type, location, time, description, tags}
    """
    qa_list = []
    name = (ent.get("name") or "").strip()
    if not name:
        return qa_list

    desc = (ent.get("description") or "").strip()
    location = (ent.get("location") or "").strip()
    time = (ent.get("time") or "").strip()
    etype = (ent.get("type") or "").lower().strip()
    tags = [t.lower() for t in ent.get("tags", []) if isinstance(t, str)]

    def add_qa(q, a):
        if a and len(qa_list) < max_questions:
            qa_list.append({"question": q, "answer": a})

    # 1) Giới thiệu chung nhưng theo kiểu “điểm đến”
    if desc:
        add_qa(
            f"{name} là điểm đến như thế nào đối với khách du lịch?",
            desc,
        )

    # 2) Vị trí – thông tin rất quan trọng với du khách
    if location:
        add_qa(
            f"{name} nằm ở đâu, thuộc khu vực nào?",
            location,
        )

    # 3) Thời điểm gắn liền (nếu có, ví dụ mùa, giai đoạn, thời kỳ)
    if time:
        add_qa(
            f"Thời điểm nào gắn liền với {name} hoặc phù hợp để tham quan?",
            time,
        )

    # 4) Câu hỏi về trải nghiệm & hoạt động (dùng description/tags trả lời)
    #    Du khách thường quan tâm: đến đó thì làm gì / có gì chơi?
    if desc:
        add_qa(
            f"Khi đến {name}, du khách thường có những trải nghiệm hoặc hoạt động nào nổi bật?",
            desc,
        )

    # 5) Tuỳ loại điểm đến, đặt câu hỏi chuyên hơn

    # Bãi biển / đảo / cảnh quan thiên nhiên
    is_beach_or_island = any(
        kw in etype
        for kw in ["beach", "bãi biển", "island", "đảo", "cảnh quan", "thiên nhiên"]
    ) or any("bãi biển" in t or "biển" in t or "đảo" in t for t in tags)

    # Di tích, văn hoá, tôn giáo
    is_cultural_site = any(
        kw in etype
        for kw in ["cultural", "cultural_site", "heritage", "di sản", "chùa", "đền", "đình", "nhà thờ", "bảo tàng"]
    ) or any(
        "di sản" in t or "văn hóa" in t or "lịch sử" in t or "tín ngưỡng" in t
        for t in tags
    )

    # Thành phố / khu du lịch tổng hợp
    is_city_or_area = any(
        kw in etype
        for kw in ["city", "thành phố", "khu du lịch", "khu nghỉ dưỡng"]
    ) or any(
        "thành phố" in t or "khu du lịch" in t or "khu nghỉ dưỡng" in t
        for t in tags
    )

    # 5a) Bãi biển / đảo
    if is_beach_or_island and desc:
        add_qa(
            f"{name} phù hợp với những loại hình du lịch biển nào (nghỉ dưỡng, khám phá, thể thao nước...)?",
            desc,
        )

    # 5b) Di tích / văn hoá
    if is_cultural_site and desc:
        add_qa(
            f"{name} có ý nghĩa văn hoá – lịch sử gì đặc biệt thu hút du khách?",
            desc,
        )

    # 5c) Thành phố / khu du lịch
    if is_city_or_area and desc:
        add_qa(
            f"Những điểm nhấn chính trong hành trình tham quan tại {name} là gì?",
            desc,
        )

    # 6) Nếu có tags, thêm 1 câu hỏi tóm tắt “phù hợp với du khách nào”
    if tags:
        add_qa(
            f"{name} phù hợp với những nhóm khách du lịch nào hoặc loại hình trải nghiệm nào?",
            "Một số từ khoá nổi bật: " + ", ".join(sorted(set(ent.get("tags", [])))),
        )

    return qa_list


def create_qa_from_event(ev: dict, max_questions: int = 5):
    """
    Tạo list Q&A cho 1 sự kiện (lễ hội / sự kiện văn hoá – du lịch...),
    tập trung vào điều du khách muốn biết.
    ev: {id, name, time, location, participants, description, tags}
    """
    qa_list = []
    name = (ev.get("name") or "").strip()
    if not name:
        return qa_list

    desc = (ev.get("description") or "").strip()
    location = (ev.get("location") or "").strip()
    time = (ev.get("time") or "").strip()
    tags = [t.lower() for t in ev.get("tags", []) if isinstance(t, str)]

    def add_qa(q, a):
        if a and len(qa_list) < max_questions:
            qa_list.append({"question": q, "answer": a})

    # Lễ hội / event là gì?
    if desc:
        add_qa(
            f"{name} là sự kiện/lễ hội như thế nào?",
            desc,
        )

    # Thời gian tổ chức – cực kỳ quan trọng với khách
    if time:
        add_qa(
            f"Lễ hội/sự kiện {name} thường diễn ra vào thời gian nào?",
            time,
        )

    # Địa điểm tổ chức
    if location:
        add_qa(
            f"Lễ hội/sự kiện {name} được tổ chức ở đâu?",
            location,
        )

    # Hoạt động chính trong lễ hội / event
    if desc:
        add_qa(
            f"Những hoạt động chính diễn ra trong lễ hội/sự kiện {name} là gì?",
            desc,
        )

    # Nếu tags có gợi ý (“ẩm thực”, “tâm linh”, “âm nhạc”...)
    if tags:
        add_qa(
            f"Lễ hội/sự kiện {name} phù hợp với những loại hình trải nghiệm văn hoá – du lịch nào?",
            "Một số từ khoá nổi bật: " + ", ".join(sorted(set(ev.get("tags", [])))),
        )

    return qa_list

````

````python
def is_culture_history_chunk(text: str) -> bool:
    """
    Trả về True nếu chunk có vẻ liên quan tới Văn hoá / Lịch sử / Di tích.
    Bạn có thể chỉnh lại danh sách keyword bên dưới cho sát nội dung PDF.
    """
    t = text.lower()

    # Từ khoá về văn hoá – lịch sử – di tích – danh thắng
    keywords_any = [
        "văn hóa", "văn hoá",
        "lịch sử", "lịch sử",
        "di tích", "di tích",
        "di sản", "di sản",
        "di sản văn hóa", "di sản văn hóa",
        "di tích lịch sử",
        "di sản thế giới",
        "danh lam thắng cảnh",
        "di tích quốc gia",
        "khu du lịch",
        "khu du lịch",
        "khu di tích",
        "nhà thờ", "chùa", "đình", "miếu",
        "bảo tàng", "bảo tàng",
        "phố cổ", "phố cổ",
        "thành cổ", "thành cổ",
        "triều đại", "triều đại",
        "thế kỷ", "thế kỷ",
        "thời kỳ", "thời kỳ",
        "kháng chiến", "kháng chiến",
    ]

    return any(kw in t for kw in keywords_any)


# Lọc chunks
filtered_chunks = [c for c in chunks if is_culture_history_chunk(c)]
print(f"Tổng số chunks ban đầu: {len(chunks)}")
print(f"Số chunks liên quan văn hoá / lịch sử: {len(filtered_chunks)}")

# Xem thử 1–2 chunk đầu sau khi lọc
for i, ch in enumerate(filtered_chunks[:2], start=1):
    print(f"\n--- Chunk {i} (rút gọn 500 ký tự) ---")
    print(ch[:500])
````

````python
import time
from google.genai.errors import ServerError  # nếu chưa có thì import

def safe_extract_json_with_gemini(text, max_retries=3, delay=2.0):
    for attempt in range(1, max_retries+1):
        try:
            return extract_json_with_gemini(text)
        except ServerError as e:
            if "UNAVAILABLE" in str(e) or "503" in str(e):
                print(f"  [WARN] Gemini 503 (model overloaded), thử lại {attempt}/{max_retries}")
                if attempt < max_retries:
                    time.sleep(delay)
                    continue
                else:
                    print("  [ERROR] Hết số lần retry, bỏ chunk này.")
                    return {}
            else:
                print(f"  [ERROR] Lỗi server khác từ Gemini: {e}")
                return {}
        except Exception as e:
            print(f"  [ERROR] Lỗi không mong đợi khi gọi Gemini: {e}")
            return {}


raw_records = []
final_entities = []
final_events = []
invalid_records = []

num_chunks = min(10, len(filtered_chunks))   # demo 10 chunk, ổn rồi thì dùng len(filtered_chunks)

for i in range(num_chunks):
    print(f"=== Xử lý chunk (văn hoá/lịch sử) {i+1}/{num_chunks} ===")
    record = safe_extract_json_with_gemini(filtered_chunks[i])
    raw_records.append(record)

    if validate_record_structure(record):
        print("  -> JSON HỢP LỆ (cho vào Final Data).")
        final_entities.extend(record.get("entities", []))
        final_events.extend(record.get("events", []))
    else:
        print("  -> JSON KHÔNG HỢP LỆ (đưa vào Invalid để Tester chỉnh).")
        invalid_records.append(record)

print("\n[SUMMARY]")
print("  Raw records:", len(raw_records))
print("  Entities hợp lệ:", len(final_entities))
print("  Events hợp lệ:", len(final_events))
print("  Records invalid:", len(invalid_records))
````
